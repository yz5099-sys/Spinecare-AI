## 1.0 Markdown Syntax

### 1.1 Headings

By using hashtags,  you can easily organize your notes's headnings:  
`#`: A title  
`##`: Section heading  
`###`: Subsection heading  
`####`: Deeper headning

### 1.2 Paragraphs

Every time you want to start a new paragrah, please type two spaces and then type the `Enter`. This is rule in markdown to start a new pararaph. Without this, Markdown will glue all paragraphs together.

### 1.3 List

In Markdown, it is very simple to create bullet list by using `-`, `*`, `+`, or `1. `.

Example 1: using `-`.  
The Top 3 movies on IMDB are:  
- The Shawshank Redemption  
- The Godfather  
- The Dark Knight    

Example 2: using `*`.  
The Top 3 movies on IMDB are:  
* The Shawshank Redemption  
* The Godfather  
* The Dark Knight  

Example 3: using `+`.  
The Top 3 movies on IMDB are:  
+ The Shawshank Redemption  
+ The Godfather  
+ The Dark Knight  

Example 4: using `1. `.  
The Top 3 movies on IMDB are:  
1. The Shawshank Redemption  
1. The Godfather  
1. The Dark Knight

#### Exercise B2 — Totals by hand
**Difficulty:** ★★☆☆☆  **Points:** 5
**Skills:** accumulator, loop + `if`

A restaurant's wine-by-the-glass prices: `prices = [12.5, 9.99, 15.75, 21.0, 7.25]`.

1. With an accumulator loop (no `sum()`), compute and print the **total** and the **mean** price, the mean rounded to 2 decimals.
2. With a loop and an `if`, **count** how many glasses cost more than $10 and print the count.

My original answer:

````python
# Your code here
prices = [12.5, 9.99, 15.75, 21.0, 7.25]

total = 0
for i in prices:
  total = total + i

count = len(prices)
mean = round(total / count,2)

print(f"The total price is {total}, and the mean is {mean:.2f}.")
````

I used AI (Gemini/GPT/Claude) in checking/refining/revising/debugging my answer:

````python
# My revised code here
prices = [12.5, 9.99, 15.75, 21.0, 7.25]

# 1. Compute total and mean using an accumulator loop
total = 0.0
for p in prices:
    total += p

mean = round(total / len(prices), 2)
print(f"The total price is {total:.2f}, and the mean is {mean:.2f}.")

# 2. Count glasses costing more than $10
count_over_10 = 0
for p in prices:
    if p > 10:
        count_over_10 += 1

print(f"Number of glasses costing more than $10: {count_over_10}")
````
